# File Input and Output with Arrays

Mostly perople use pandas for data load <br>
numpy.save and numpy.load are two functions

```python
import numpy as np
```

```python
arr = np.arange(11)
```

```python
arr
```

```python
np.save("Some_array",arr)
```

```python
np.load("some_array.npy")
```

```python
np.savez("some_array.npz", a=arr, b=arr)
```

```python
np.load("some_array.npz")
```
