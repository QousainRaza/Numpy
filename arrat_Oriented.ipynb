## Expressing Conditional Logic as Array Operations

```python
xarr = np.array([1.1,1.2,1.3,1.4,1.5])
yarr = np.array([2.1,2.2,2.3,2.4,2.5])
conda = np.array([True,False,False,True,True])
```

we wanted to take a value from xarr whenever the corresponding value in
cond is True, and otherwise take the value from yarr. A list comprehension doing
this might look like:

```python
result = [(x if c else y )
          for x, y, c in zip(xarr,yarr,conda)]
```

```python
result
```

This has multiple issues: <br>
1. It will not fast
2. It will not work with Mult-dimnesional Array

```python
## numpy.where do this with a single function

result = np.where(xarr,yarr,conda)
```

```python
result
```

## Mathematical and Statistical Methods

```python
rng = np.random.default_rng(seed=1234)
```

```python
arr = rng.standard_normal((5,4))
```

```python
arr
```

```python
arr.mean()
```

```python
np.mean(arr)
```

```python
arr.mean(axis=1)  # MEANS find mean of each row
```

```python
arr.sum(axis=0) # find sum down the row
```

```python
arr = np.array([[1,2,3],[3,5,6],[7,8,9]])
arr
```

```python
arr.cumsum(axis=0)
```

```python
arr.cumsum(axis=1)
```
